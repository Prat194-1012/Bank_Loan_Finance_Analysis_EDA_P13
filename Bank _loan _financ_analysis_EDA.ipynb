# CHARTS

### MONTHLY TRENDS BY ISSUE DATE FOR TOTAL FUNDED AMOUNT

```python
monthly_trends =(
                 df.sort_values('issue_date')
                .assign(month_name = lambda x: x['issue_date'].dt.strftime('%B %Y'))
                .groupby('month_name', sort=False)['loan_amount']
                .sum()
                .divide(1000000)
                .reset_index(name = 'loan_amount_millions')
)

plt.figure(figsize=[10,5])
plt.fill_between(monthly_trends['month_name'], monthly_trends['loan_amount_millions'], color= 'skyblue', alpha = 0.5)
plt.plot(monthly_trends['month_name'], monthly_trends['loan_amount_millions'], color= 'blue', linewidth = 2)

for i, rows in monthly_trends.iterrows():
    plt.text(i,rows['loan_amount_millions'] + 0.1, f"{rows['loan_amount_millions']:.2f}",
             ha = 'center', va = 'bottom', fontsize = 9, rotation = 0, color='Black')

plt.title("TOTAL FUNDED AMOUNT BY MONTH", fontsize = 14)
plt.xlabel('Months')
plt.ylabel('Total Funded Amount')
plt.xticks(ticks = range(len(monthly_trends)), labels=monthly_trends['month_name'], rotation = 45)
plt.grid(True, linestyle ='-', alpha=0.6)
plt.tight_layout()
plt.show()
    
```

### MONTHLY TRENDS BY ISSUE DATE FOR TOTAL RECIEVED AMOUNT

```python
monthly_trends =(
                 df.sort_values('issue_date')
                .assign(month_name = lambda x: x['issue_date'].dt.strftime('%B %Y'))
                .groupby('month_name', sort=False)['total_payment']
                .sum()
                .divide(1000000)
                .reset_index(name = 'recieved_amount_millions')
)

plt.figure(figsize=[10,5])
plt.fill_between(monthly_trends['month_name'], monthly_trends['recieved_amount_millions'], color= 'lightgreen', alpha = 0.5)
plt.plot(monthly_trends['month_name'], monthly_trends['recieved_amount_millions'], color= 'blue', linewidth = 2)

for i, rows in monthly_trends.iterrows():
    plt.text(i,rows['recieved_amount_millions'] + 0.1, f"{rows['recieved_amount_millions']:.2f}",
             ha = 'center', va = 'bottom', fontsize = 9, rotation = 0, color='Black')

plt.title("TOTAL RECIEVED AMOUNT BY MONTH", fontsize = 14)
plt.xlabel('Months')
plt.ylabel('Total recieved Amount')
plt.xticks(ticks = range(len(monthly_trends)), labels=monthly_trends['month_name'], rotation = 45)
plt.grid(True, linestyle ='-', alpha=0.6)
plt.tight_layout()
plt.show()
    
```

### MONTHLY TRENDS BY ISSUE DATE FOR TOTAL LOAN APLLICATIONS

```python
monthly_trends =(
                 df.sort_values('issue_date')
                .assign(month_name = lambda x: x['issue_date'].dt.strftime('%B %Y'))
                .groupby('month_name', sort=False)['application_type']
                .count()
                .reset_index(name = 'loan_applications_count')
)

plt.figure(figsize=[10,5])
plt.fill_between(monthly_trends['month_name'], monthly_trends['loan_applications_count'], color= 'lightpink', alpha = 0.5)
plt.plot(monthly_trends['month_name'], monthly_trends['loan_applications_count'], color= 'green', linewidth = 2)

for i, rows in monthly_trends.iterrows():
    plt.text(i,rows['loan_applications_count'] + 0.1, f"{rows['loan_applications_count']:}",
             ha = 'center', va = 'bottom', fontsize = 9, rotation = 0, color='Black')

plt.title("TOTAL LOAN APPLICATION BY MONTH", fontsize = 14)
plt.xlabel('Months')
plt.ylabel('Total Loan Application')
plt.xticks(ticks = range(len(monthly_trends)), labels=monthly_trends['month_name'], rotation = 45)
plt.grid(True, linestyle ='-', alpha=0.6)
plt.tight_layout()
plt.show()
```

### REGIONAL ANALYSIS BY STATE FOR TOTAL FUNDED AMOUNT

```python
state_funding = df.groupby('address_state')['loan_amount'].sum().sort_values(ascending = True)
state_funding_thoushand = state_funding/1000 

plt.figure(figsize=[10,8])
bars = plt.barh(state_funding_thoushand.index,state_funding_thoushand.values, color='orange')

for bar in bars:
    width = bar.get_width()
    plt.text(width + 10, bar.get_y() + bar.get_height()/2,
             f"{width:,.0f}k",va ='center',fontsize = 9)
    
plt.title("TOTAL FUNDED AMOUNT BY STATE(in $ thoushand)")
plt.xlabel('funded amount ($ thoushand)')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

### REGIONAL ANALYSIS BY STATE FOR TOTAL AMOUNT RECIEVED

```python
state_funding = df.groupby('address_state')['total_payment'].sum().sort_values(ascending = True)
state_total_amount_thoushand = state_funding/1000 

plt.figure(figsize=[10,8])
bars = plt.barh(state_total_amount_thoushand.index,state_total_amount_thoushand.values, color='darkblue')

for bar in bars:
    width = bar.get_width()
    plt.text(width + 10, bar.get_y() + bar.get_height()/2,
             f"{width:,.0f}k",va ='center',fontsize = 9)
    
plt.title("TOTAL AMOUNT RECIEVED BY STATE(in $ thoushand)")
plt.xlabel('Total amount ($ thoushand)')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

### LOAN TERM ANALYSIS BY TOTAL FUNDED AMOUNT

```python
term_funding_millions = df.groupby('term')['loan_amount'].sum()/1000000

plt.figure(figsize=[5,5])
plt.pie(
    term_funding_millions,
    labels = term_funding_millions.index,
    autopct = lambda p: f"{p:.1f}%\n${p*sum(term_funding_millions)/100:.0f}M",
    startangle = 90,
    wedgeprops = {'width': 0.4}
)
plt.gca().add_artist(plt.Circle((0,0),0.70, color='white'))
plt.title("Total funded amount by terms (in $ thoushands)")
plt.show()
```

### HOME OWNERSHIP BY TOTAL FUNDED AMOUNT 

```python
home_funding= df.groupby('home_ownership')['loan_amount'].sum().reset_index()
home_funding['loan amount millions'] = home_funding['loan_amount']/1000000
fig= px.treemap(
    home_funding,
    path=['home_ownership'],
    values='loan amount millions',
    color='loan amount millions',
    color_continuous_scale='Blues',
    title='Total funded amount by home ownership ($ millions)'
)
fig.show()
```

### HOME OWNERSHIP BY TOTAL RECIEVED AMOUNT 

```python
home_funding= df.groupby('home_ownership')['total_payment'].sum().reset_index()
home_funding['total amount millions'] = home_funding['total_payment']/1000000
fig= px.treemap(
    home_funding,
    path=['home_ownership'],
    values='total amount millions',
    color='total amount millions',
    color_continuous_scale='reds',
    title='Total amount recieved by home ownership ($ millions)'
)
fig.show()
```

### HOME OWNERSHIP BY TOTAL LOAN APPLICATION
